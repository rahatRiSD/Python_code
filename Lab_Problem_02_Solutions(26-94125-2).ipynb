## Part (a): Strings

**Approach**

- A string is a sequence of characters, so each character has a position (index) starting at `0`.
- `message[0]` gives the first character, `C`.
- `message[0:8]` is a slice from index 0 up to (but not including) index 8, which is the word `Computer`.
- `message[-7:]` uses a negative index: `-7` counts 7 characters back from the end, and leaving the end blank means "go to the end", giving `Science`.
- Passing several values to one `print()` separated by commas prints them on one line with a space between each, so `"|"` becomes a visible separator.

For `course`:

- `strip()` returns a copy with the leading and trailing spaces removed (the original string is unchanged, because strings are immutable).
- `upper()` returns the uppercase version.
- `replace("Python", "Java")` returns a copy with every `Python` swapped for `Java`.
- The `in` operator checks whether one string appears inside another and gives a Boolean (`True`/`False`).

```python
print("=== 1. Strings ===")

message = "Computer Science"
print(message[0], "|", message[0:8], "|", message[-7:])

course = "  Python Programming  "
clean = course.strip()
print(clean.upper())
print(clean.replace("Python", "Java"))
print("Programming" in clean)
```

## Part (b): Lists, part 1

**Approach**

- A list is an ordered, changeable (mutable) collection written with square brackets.
- `append(85)` adds `85` to the end of the list, changing the list in place.
- `remove(60)` deletes the first item equal to `60`.
- After both changes the list is `[75, 90, 85]`.
- `max()` returns the largest value in the list, which is `90`.
- The heading string starts with `\n`, which prints the required blank line before it.

```python
print("\n=== 2. Lists ===")

scores = [60, 75, 90]
scores.append(85)
scores.remove(60)
print(scores, "Highest:", max(scores))
```

## Part (c): Lists, part 2

**Approach**

- A `for` loop visits each item of `marks` in order; the loop variable `mark` holds the current item, and the indented `print` runs once per item.
- The average is calculated rather than hard-coded: `sum(marks)` adds all the values (300) and `len(marks)` counts them (4).
- In Python 3, `/` always gives a decimal result, so the average prints as `75.0`.
- The average is printed after the loop (not indented), so it prints only once.

```python
marks = [70, 80, 90, 60]
for mark in marks:
    print("Mark:", mark)
print("Average:", sum(marks) / len(marks))
```

## Part (d): Tuples

**Approach**

- A tuple is like a list but immutable (it cannot be changed after it is created); it is written with round brackets.
- Indexing works the same way as for lists: `subjects[0]` is the first element and `subjects[-1]` is the last one.
- `len(subjects)` gives the number of elements, `3`.
- **Tuple unpacking** assigns each element of the tuple to its own variable in one statement: `student_id, name, cgpa = student`. The number of variables must match the number of elements.
- The values are then printed in the order the task asks for: name, ID, CGPA.

```python
print("\n=== 3. Tuples ===")

subjects = ("Python", "Algorithms", "Database")
print(subjects[0], subjects[-1], "Total:", len(subjects))

student = (101, "Rahim", 3.75)
student_id, name, cgpa = student
print("Name:", name, "ID:", student_id, "CGPA:", cgpa)
```

## Part (e): Sets

**Approach**

- A set stores only unique values, so writing `10` and `20` twice still leaves just one of each: `{10, 20, 30}`.
- `add(40)` inserts a new element.
- Sets have no fixed order, so `sorted()` is used to return the elements as a sorted list, making the output predictable.
- `20 in numbers` tests membership and gives `True`.

Set operators:

| Operator | Meaning | Result for `a` and `b` |
|---|---|---|
| `a \| b` | union: everything in either set | `[1, 2, 3, 4, 5, 6]` |
| `a & b` | intersection: only what is in both | `[3, 4]` |
| `a - b` | difference: in `a` but not in `b` | `[1, 2]` |

```python
print("\n=== 4. Sets ===")

numbers = {10, 20, 10, 30, 20}
numbers.add(40)
print(sorted(numbers), "Contains 20:", 20 in numbers)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union:", sorted(a | b))
print("Intersection:", sorted(a & b))
print("Only A:", sorted(a - b))
```

## Part (f): Dictionaries

**Approach**

- A dictionary stores **key: value** pairs inside curly brackets, and a value is looked up by its key instead of by position.
- `student["name"]` returns `Rina`.
- Assigning to an existing key, `student["cgpa"] = 3.75`, replaces its old value.
- `items()` returns each key–value pair together; in the `for` loop they are unpacked into `key` and `value`.
- Printing `key, ":", value` with commas puts a space on each side of the colon, giving the required `name : Notebook` format.

```python
print("\n=== 5. Dictionaries ===")

student = {"name": "Rina", "id": 102, "cgpa": 3.50}
print(student["name"])
student["cgpa"] = 3.75
print("Updated CGPA:", student["cgpa"])

product = {"name": "Notebook", "price": 80, "stock": 12}
for key, value in product.items():
    print(key, ":", value)
```

## Part (g): User-defined functions

**Approach**

- `def` defines a function: a named block of code that runs only when it is called.
- `greet()` has no parameters and simply prints a message, so calling it twice prints the message twice.
- `display_square(number)` takes one parameter. The value passed in the call (`4`, then `7`) is stored in `number`, and the function prints `number * number`.
- As required, `display_square` **prints** the result itself and does not return anything.

```python
print("\n=== 6. User-defined Functions ===")

def greet():
    print("Welcome to Python!")

greet()
greet()

def display_square(number):
    print("Square:", number * number)

display_square(4)
display_square(7)
```

## Part (h): Function parameters

**Approach**

- **Positional parameters:** in `introduce("Rahim", "CSE")`, the first value goes to `name` and the second to `department`, matched by position.
- **Default parameter:** `semester="Fall"` gives `semester` a value to use when the caller leaves it out, so `course_info("Python")` prints `Python - Fall`.
- **Keyword arguments:** naming each parameter in the call (`semester="Spring", course="Algorithms"`) lets the arguments be passed in any order; Python matches them by name, so the output is still `Algorithms - Spring`.

```python
print("\n=== 7. Function Parameters ===")

def introduce(name, department):
    print(name, "studies", department)

introduce("Rahim", "CSE")
introduce("Nila", "EEE")

def course_info(course, semester="Fall"):
    print(course, "-", semester)

course_info("Python")
course_info(semester="Spring", course="Algorithms")
```

## Part (i): Return statement, lambda, recursion

### 1. Return statement

- `return` sends a value back to where the function was called, so the result can be printed or stored. `add(10, 15)` gives back `25`.
- Writing `return a + b, a * b` returns both values packed in a tuple, `(9, 20)`, which is then unpacked into `sum_result` and `product`.

### 2. Lambda functions

- A `lambda` is a small, unnamed function written in one expression. `lambda x: x * x` does the same as a `def` function that returns `x * x`.
- `sorted()` accepts a `key` function that decides what each item is compared by. `lambda item: item[1]` picks the score (the second element of each tuple), and `reverse=True` sorts from highest to lowest.

### 3. Recursion

- A recursive function calls itself on a smaller problem, and it needs a **base case** to stop.
- `factorial(n)` uses the rule n! = n × (n − 1)!, with the base case `factorial(0) = 1`. So 5! = 5 × 4 × 3 × 2 × 1 = 120.
- `countdown(n)` prints `Calling:` on the way down (3, 2, 1), reaches the base case at 0, then each waiting call finishes and prints `Returning:` on the way back up (1, 2, 3). This shows how the calls stack up and then unwind in reverse order.

```python
print("\n=== 8. Return Statement ===")

def add(a, b):
    return a + b

print("Sum:", add(10, 15))

def calculations(a, b):
    return a + b, a * b

sum_result, product = calculations(4, 5)
print("Sum:", sum_result)
print("Product:", product)

print("\n=== 9. Lambda Functions ===")

square = lambda x: x * x
print(square(6))
print(square(9))

students = [("Rina", 80), ("Karim", 95), ("Mina", 75)]
ranked = sorted(students, key=lambda item: item[1], reverse=True)
print(ranked)

print("\n=== 10. Recursion ===")

def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print("5! =", factorial(5))

def countdown(n):
    if n == 0:
        print("Base case")
        return
    print("Calling:", n)
    countdown(n - 1)
    print("Returning:", n)

countdown(3)
```

```python
# Exercise 3: Random sampling
import random
from collections import Counter

rng = random.Random(11)

picked = rng.sample(range(1, 11), 4)
print("Four distinct values:", picked)

rolls = [rng.randint(1, 6) for _ in range(600)]
counts = Counter(rolls)
print("Face | Proportion")
for face in range(1, 7):
    print(f"  {face}  | {counts[face] / 600:.3f}")

```

```python
# Exercise 4: Exception flow
def attempt(raw):
    try:
        result = 100 / int(raw)
    except ValueError:
        print(f"  [except ValueError] '{raw}' is not an integer")
    except ZeroDivisionError:
        print(f"  [except ZeroDivisionError] cannot divide by zero")
    else:
        print(f"  [else] 100 / {raw} = {result}")
    finally:
        print(f"  [finally] done with '{raw}'")

for value in ("4", "0", "abc"):
    print(f"attempt({value!r}):")
    attempt(value)
```

```python
# Exercise 5: Raising exceptions
def parse_lab_mark(raw):
    mark = int(raw)
    if not 0 <= mark <= 100:
        raise ValueError(f"mark outside 0-100: {mark}")
    return mark

for raw in ("63", "120", "n/a"):
    try:
        mark = parse_lab_mark(raw)
    except ValueError as err:
        print(f"Invalid: {raw!r} -> {err}")
    else:
        print(f"Accepted: {mark}")
```

```python
# Exercise 8: CSV filtering
import csv

# Create sample input data
with open("results.csv", "w", newline="", encoding="utf-8") as f:
    w = csv.writer(f)
    w.writerow(["name", "mark"])
    w.writerows([["Rahim", 82], ["Mina", 74], ["Karim", 75], ["Sadia", 91], ["Tanvir", 40]])

def filter_passed(src, dst, threshold=75):
    with open(src, "r", newline="", encoding="utf-8") as fin:
        reader = csv.DictReader(fin)
        if reader.fieldnames is None or "mark" not in reader.fieldnames:
            print(f"Error: '{src}' has no 'mark' column; nothing written.")
            return
        kept = 0
        with open(dst, "w", newline="", encoding="utf-8") as fout:
            writer = csv.DictWriter(fout, fieldnames=reader.fieldnames)
            writer.writeheader()
            for row in reader:
                if int(row["mark"]) >= threshold:
                    writer.writerow(row)
                    kept += 1
    print(f"Kept {kept} record(s) in {dst}")

filter_passed("results.csv", "passed.csv")
with open("passed.csv", encoding="utf-8") as f:
    print(f.read())

with open("bad_results.csv", "w", newline="", encoding="utf-8") as f:
    csv.writer(f).writerows([["name", "score"], ["Rahim", 82]])
filter_passed("bad_results.csv", "bad_passed.csv")
```

```python
# Exercise 10: Integrated lab
import csv
import re

ID_PATTERN = re.compile(r"[A-Z]{3}-\d{4}-\d{3}", re.ASCII)
REQUIRED = {"student_id", "mark"}

def process_marks(path):
    try:
        f = open(path, "r", newline="", encoding="utf-8")
    except FileNotFoundError:
        print(f"Error: file '{path}' not found.")
        return
    with f:
        reader = csv.DictReader(f)
        if reader.fieldnames is None:
            print(f"Error: '{path}' is empty.")
            return
        missing = REQUIRED - set(reader.fieldnames)
        if missing:
            print(f"Error: missing column(s): {', '.join(sorted(missing))}")
            return

        marks, rejected = [], 0
        for row in reader:
            sid = (row["student_id"] or "").strip()
            raw = (row["mark"] or "").strip()
            if not ID_PATTERN.fullmatch(sid):
                print(f"Line {reader.line_num}: invalid student ID '{sid}'")
                rejected += 1
                continue
            try:
                mark = int(raw)
            except ValueError:
                print(f"Line {reader.line_num}: mark '{raw}' is not an integer")
                rejected += 1
                continue
            if not 0 <= mark <= 100:
                print(f"Line {reader.line_num}: mark {mark} outside 0-100")
                rejected += 1
                continue
            marks.append(mark)

    print("Accepted:", len(marks))
    print("Rejected:", rejected)
    if marks:
        print(f"Mean: {sum(marks) / len(marks):.2f}")
        print("Highest:", max(marks))
    else:
        print("Mean: N/A (no valid rows)")
        print("Highest: N/A")

if __name__ == "__main__":
    # Sample data
    with open("marks.csv", "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f)
        w.writerow(["student_id", "mark"])
        w.writerows([
            ["EEE-2025-017", "88"],
            ["CSE-2024-101", "72"],
            ["eee-2025-018", "90"],
            ["MEC-2023-045", "abc"],
            ["EEE-2025-020", "105"],
            ["CSE-2024-102", "65"],
        ])
    with open("header_only.csv", "w", newline="", encoding="utf-8") as f:
        csv.writer(f).writerow(["student_id", "mark"])

    print("=== marks.csv ===")
    process_marks("marks.csv")
    print("\n=== header_only.csv ===")
    process_marks("header_only.csv")
    print("\n=== missing.csv ===")
    process_marks("missing.csv")
```

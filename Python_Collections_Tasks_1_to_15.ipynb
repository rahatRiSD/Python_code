## Task 1 — Create a list and access the 2nd element
**Concept:** Python indexing starts at **0**, so the 2nd element is index `1`.

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[1])   # banana
```

## Task 2 — Slicing and `append()`
**Concept:** `list[start:stop]` includes `start`, excludes `stop`. So `[1:4]` gives indices 1,2,3 → `[2, 3, 4]`.

```python
numbers = [1, 2, 3, 4, 5]
print(numbers[1:4])      # [2, 3, 4]
numbers.append(6)        # adds to the END, in place
print(numbers)           # [1, 2, 3, 4, 5, 6]
```

## Task 3 — `list2 = list1` vs `list2 = list1.copy()`
**Explanation:**
- `list2 = list1` → **no new list**; both names point to the *same object* (alias). Changing one changes the other.
- `list2 = list1.copy()` → creates a **new, independent (shallow) copy**. Changes to `list1` don't affect `list2`.

```python
# Scenario A: alias (same object)
list1 = [1, 2, 3]
list2 = list1
list1.append(99)
print("Alias  -> list1:", list1, "| list2:", list2, "| same object?", list1 is list2)

# Scenario B: copy (independent object)
list1 = [1, 2, 3]
list2 = list1.copy()
list1.append(99)
print("Copy   -> list1:", list1, "| list2:", list2, "| same object?", list1 is list2)
```

## Task 4 — Tuple + negative indexing
**Concept:** `-1` = last element, `-2` = second last, etc.

```python
colors = ("red", "green", "blue")
print(colors[-1])   # blue
```

## Task 5 — Concatenate tuples; why no `.append()`?
**Explanation:** Tuples are **immutable** — they have no `.append()` because that would modify the tuple in place. Instead, `+` builds a **new tuple**.

```python
numbers = (1, 2, 3, 4)
new_numbers = numbers + (5, 6)
print(new_numbers)          # (1, 2, 3, 4, 5, 6)

try:
    numbers.append(5)
except AttributeError as e:
    print("Error:", e)
```

## Task 6 — Why tuples are immutable / read-only data
**Explanation:** Once created, a tuple's elements cannot be added, removed or reassigned. This makes tuples:
- **Safe** for constant data (e.g., coordinates, RGB values, config) — no accidental changes.
- **Hashable** (if elements are hashable) → can be dict keys / set members.
- Slightly **faster and lighter** than lists.

Trying `colors[0] = "yellow"` raises `TypeError`.

```python
colors = ("red", "green", "blue")
try:
    colors[0] = "yellow"
except TypeError as e:
    print("TypeError:", e)

# Bonus: tuples can be dictionary keys, lists cannot
locations = {(23.7, 89.1): "Rangpur"}
print(locations[(23.7, 89.1)])
```

## Task 7 — Set and uniqueness
**Explanation:** A set is a collection of **unique** items (implemented with a hash table). Duplicate values are discarded automatically when the set is created.

```python
my_set = {1, 2, 2, 3, 3, 4}
print(my_set)    # {1, 2, 3, 4}  -> duplicates removed
```

## Task 8 — Union, Intersection, Difference

```python
a = {1, 2, 3}
b = {3, 4, 5}
print("Union        :", a | b)    # same as a.union(b)        -> {1,2,3,4,5}
print("Intersection :", a & b)    # same as a.intersection(b) -> {3}
print("Difference   :", a - b)    # same as a.difference(b)   -> {1,2}
```

## Task 9 — Remove duplicates and sort
**Concept:** `set()` removes duplicates in O(n); `sorted()` returns a sorted **list** (and sorts any iterable).

```python
items = [1, 2, 2, 3, 4, 4, 5]
unique_sorted = sorted(set(items))
print(unique_sorted)    # [1, 2, 3, 4, 5]
```

## Task 10 — `append()` vs `extend()`
- `append(x)` adds **x as one single element** (even if x is a list).
- `extend(iterable)` adds **each element** of the iterable individually.

```python
a = [1, 2]; a.append([3, 4]);  print("append :", a)   # [1, 2, [3, 4]]
b = [1, 2]; b.extend([3, 4]);  print("extend :", b)   # [1, 2, 3, 4]
```

## Task 11 — `remove()` vs `discard()` (sets)
| Method | Element exists | Element missing |
|---|---|---|
| `remove(x)` | removes it | **raises `KeyError`** |
| `discard(x)` | removes it | **does nothing** (no error) |

```python
s = {1, 2, 3}

s.discard(10)                 # no error
print("discard(10) OK ->", s)

try:
    s.remove(10)
except KeyError as e:
    print("remove(10) KeyError:", e)
```

## Task 12 — Common elements of a list and a tuple
**Concept:** Convert both to sets, then use intersection (`&`).

```python
my_list  = [1, 2, 3, 2, 1]
my_tuple = (3, 4, 5)
common = set(my_list) & set(my_tuple)
print(common)    # {3}
```

## Task 14 — Built-in functions: `len()`, `sum()`, `sorted()`
`sorted()` returns a **new** sorted list and leaves the original unchanged.

```python
data = [10, 25, 5, 40]
print("Length :", len(data))      # 4
print("Sum    :", sum(data))      # 80
print("Sorted :", sorted(data))   # [5, 10, 25, 40]
print("Original unchanged:", data)
```

## Task 15 — Membership test on 1,000,000 words: List or Set?
**Answer: Set.**
- `x in list` scans elements one by one → **O(n)** (up to 1,000,000 comparisons).
- `x in set` uses hashing → **O(1) average** (almost instant).

Benchmark below proves it.

```python
import time, random, string

# Build 1,000,000 random words
words = [''.join(random.choices(string.ascii_lowercase, k=8)) for _ in range(1_000_000)]
words.append("python")           # worst case for list: last position

word_list = words
word_set  = set(words)

# Time 100 lookups for each
start = time.perf_counter()
for _ in range(100): "python" in word_list
list_time = time.perf_counter() - start

start = time.perf_counter()
for _ in range(100): "python" in word_set
set_time = time.perf_counter() - start

print(f"List lookup x100: {list_time:.5f} s")
print(f"Set  lookup x100: {set_time:.7f} s")
print(f"Set is ~{list_time/set_time:,.0f}x faster")
```
